## 0) ¿Qué es recursión? (idea simple)

Una función **recursiva** es una función que **se llama a sí misma**.

Para que funcione, SIEMPRE necesita:
1) **Caso base**: condición que detiene la recursión  
2) **Paso recursivo**: reducción del problema (cada vez más pequeño)

> Si no hay caso base, la función se llama infinitamente y termina en error.

## 1) El ejemplo clásico: SUMATORIA recursiva

Queremos:
\[
S(n) = 1 + 2 + \dots + n
\]

Forma recursiva:
- **Caso base**: `S(1) = 1`
- **Paso recursivo**: `S(n) = n + S(n-1)`


```python
def sumatoria_r(n):
    # Caso base
    if n == 1:
        return 1
    # Paso recursivo
    return n + sumatoria_r(n - 1)

print(sumatoria_r(5))   # 15
print(sumatoria_r(10))  # 55
```

## 2) Traza (ver qué pasa “por dentro”)

La recursión funciona como una pila de llamadas:

Ejemplo: `sumatoria_r(4)`
- `sumatoria_r(4) = 4 + sumatoria_r(3)`
- `sumatoria_r(3) = 3 + sumatoria_r(2)`
- `sumatoria_r(2) = 2 + sumatoria_r(1)`
- `sumatoria_r(1) = 1`  ✅ caso base

Luego “vuelve hacia arriba”:
- `sumatoria_r(2) = 2 + 1 = 3`
- `sumatoria_r(3) = 3 + 3 = 6`
- `sumatoria_r(4) = 4 + 6 = 10`


```python
def sumatoria_r_con_print(n):
    # OJO: esto es solo para aprender, el curso se enfoca en return.
    print("Llamada con n =", n)

    if n == 1:
        print("Caso base, retorno 1")
        return 1

    res = n + sumatoria_r_con_print(n - 1)
    print("Retorno", res, "para n =", n)
    return res

print(sumatoria_r_con_print(4))
```

## 3) FACTORIAL recursivo

Definición:
\[
n! = 1 \cdot 2 \cdot \dots \cdot n
\]

Forma recursiva:
- **Caso base**: `0! = 1` (y también `1! = 1`)
- **Paso recursivo**: `n! = n * (n-1)!`


```python
def factorial_r(n):
    # Caso base
    if n == 0:
        return 1
    # Paso recursivo
    return n * factorial_r(n - 1)

print(factorial_r(0))  # 1
print(factorial_r(1))  # 1
print(factorial_r(5))  # 120
```

## 4) Comparación: recursión vs ciclo (mismo resultado)

En muchos problemas básicos, recursión y ciclos pueden hacer lo mismo.
La diferencia es **cómo piensas** el problema.

Abajo: factorial con `while` (sin recursión) solo para comparar.


```python
def factorial_ciclo(n):
    if n == 0:
        return 1

    resultado = 1
    i = 1
    while i <= n:
        resultado = resultado * i
        i = i + 1
    return resultado

print(factorial_ciclo(5))
```

## 5) Otro ejemplo ultra típico: POTENCIA recursiva

Queremos `base^exp` sin usar `**`.

Recursión:
- **Caso base**: `base^0 = 1`
- **Paso recursivo**: `base^exp = base * base^(exp-1)`


```python
def potencia_r(base, exp):
    if exp == 0:
        return 1
    return base * potencia_r(base, exp - 1)

print(potencia_r(2, 5))  # 32
print(potencia_r(3, 4))  # 81
```

## 6) Recursión con números: contar dígitos (básico)

Ejemplo: `contar_digitos(5002) = 4`

Idea:
- **Caso base**: si `n < 10`, tiene 1 dígito
- **Paso recursivo**: quitar el último dígito con `n // 10`


```python
def contar_digitos_r(n):
    if n < 10:
        return 1
    return 1 + contar_digitos_r(n // 10)

print(contar_digitos_r(7))       # 1
print(contar_digitos_r(5002))    # 4
print(contar_digitos_r(123456))  # 6
```

## 7) Errores comunes (MUY IMPORTANTES)

1) **No tener caso base** → recursión infinita  
2) **No reducir el problema** → también infinito  
3) Usar un caso base incorrecto (ej: `n==1` pero llamas con `n==0`)  
4) N muy grande → Python tiene límite de recursión (para el curso, usamos n chicos)

> Consejo: prueba con valores pequeños (3, 4, 5) y traza con prints si te pierdes.

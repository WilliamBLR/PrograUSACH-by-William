## 0) ¿Qué es una función?

Una función es un bloque de código con nombre que se puede reutilizar.

Idea máquina:
- **Entrada**: parámetros
- **Proceso**: código
- **Salida**: un valor con `return`


````python
# Ejemplo mínimo: devuelve un número
def devolver_cinco():
    return 5

x = devolver_cinco()
print(x)
````

## 1) `print` vs `return` (lo importante del curso)

- `print(...)` muestra en pantalla, pero **no entrega** un valor para seguir trabajando.
- `return ...` **devuelve** un valor para guardarlo en una variable y usarlo después.

Regla práctica:
> Si el resultado se necesita para cálculos, comparaciones, guardar, etc. → usa `return`.


````python
def sumar(a, b):
    return a + b

r1 = sumar(10, 7)
r2 = sumar(r1, 3)      # usamos el resultado anterior
print("r1:", r1)
print("r2:", r2)
````

## 4) Funciones con ciclos: SUMATORIA (ejercicio 1)

Definir una función:

\[
\text{sumatoria}(n) = 1 + 2 + 3 + \dots + n
\]

Ejemplo:
- `sumatoria(5) = 15`


````python
def sumatoria(n):
    total = 0
    i = 1
    while i <= n:
        total = total + i
        i = i + 1
    return total

print(sumatoria(5))
print(sumatoria(100))
````

## 5) Funciones con ciclos: FACTORIAL (ejercicio 2 - parte A)

Definir:

\[
n! = 1 \cdot 2 \cdot 3 \cdot \dots \cdot n
\]

Ejemplos:
- `0! = 1`
- `1! = 1`
- `5! = 120`


````python
def factorial(n):
    # Consideramos n >= 0
    if n == 0:
        return 1

    resultado = 1
    i = 1
    while i <= n:
        resultado = resultado * i
        i = i + 1
    return resultado

print(factorial(0))
print(factorial(1))
print(factorial(5))
````

## 6) Sumatoria de factoriales (ejercicio 2 - parte B)

Ahora combinamos funciones. Queremos:

\[
S(n) = 1! + 2! + 3! + \dots + n!
\]

La idea del curso:
1) Hacer **una función factorial(n)**  
2) Hacer **otra función** que use la anterior para sumar

Ejemplo:
- `S(4) = 1! + 2! + 3! + 4! = 1 + 2 + 6 + 24 = 33`


````python
def sumatoria_factorial(n):
    total = 0
    i = 1
    while i <= n:
        total = total + factorial(i)
        i = i + 1
    return total

print(sumatoria_factorial(4))
print(sumatoria_factorial(6))
````
